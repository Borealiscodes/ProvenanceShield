# 🕵️‍♂️ ProvenanceShield v3.0: Automated Forensic Audit & Visualization Tool
This notebook serves as an automated verification suite to validate the **cryptographic continuity** of an execution ledger and visualize state transitions over time.

```python
import json
import hashlib
import matplotlib.pyplot as plt
print("📊 Visual and cryptographic auditing dependencies loaded.")
```

### 🧬 Step 1: Generate Mock Sequential Log Stream
We generate a sequence of mock log blocks where each entry's `previous_block_hash` field points directly back to the signature of its parent to mimic real hardware-attested capture.

```python
def generate_mock_ledger():
    ledger = []
    prev_hash = "0" * 64
    
    # Simulate 5 sequential entries capturing administrative compression events
    for i in range(5):
        block_content = {
            "collection_method": "Rust Native TEE Memory Isolation Capture",
            "previous_block_hash": prev_hash,
            "telemetry_payload_hash": hashlib.sha256(f"event_data_{i}".encode()).hexdigest(),
            "state_vector": {"mandate": 1.0, "operational": 0.8 - (i * 0.2), "harm": 0.05 + (i * 0.15)}
        }
        # Enforce alphabetization for serialization stability
        raw_bytes = json.dumps(block_content, sort_keys=True).encode()
        current_hash = hashlib.sha256(raw_bytes).hexdigest()
        block_content["current_block_hash"] = current_hash
        
        ledger.append(block_content)
        prev_hash = current_hash
    return ledger

mock_log_stream = generate_mock_ledger()
print(f"📦 Compiled {len(mock_log_stream)} sequential telemetry tracking blocks into memory.")
```

### ⛓️ Step 2: The Cryptographic Continuity Audit
We loop through the ledger entries to verify the **N → N-1 link**. If an adversary has cut out or altered a row, the sequence verification will break instantly.

```python
def verify_log_chain(ledger):
    print("🔍 Initiating full sequence verification check...")
    expected_prev = "0" * 64
    
    for idx, block in enumerate(ledger):
        # Assert continuity with preceding signature
        if block["previous_block_hash"] != expected_prev:
            print(f"❌ [CHAIN BROKEN] Cryptographic discontinuity detected at entry block index: {idx}")
            return False
        
        # Recalculate block signature to confirm internal text variables weren't altered
        temp_block = block.copy()
        target_hash = temp_block.pop("current_block_hash")
        
        raw_bytes = json.dumps(temp_block, sort_keys=True).encode()
        recalculated_hash = hashlib.sha256(raw_bytes).hexdigest()
        
        if recalculated_hash != target_hash:
            print(f"❌ [DATA TAMPERING] Block {idx} content checksum mismatch!")
            return False
            
        expected_prev = target_hash
        
    print("✅ [VERIFICATION SUCCESS] The sequential hash chain is pristine and court-ready.")
    return True

verify_log_chain(mock_log_stream)
```

### 📉 Step 3: Visualizing the State Vector Phase-Space Trajectory
We extract the metrics embedded in our hardware logs and plot how our **Snowman spatial layers** deform as tracking pressures mount.

```python
steps = []
operational_radius = []
harm_radius = []

for idx, block in enumerate(mock_log_stream):
    steps.append(f"Step {idx}")
    operational_radius.append(block["state_vector"]["operational"])
    harm_radius.append(block["state_vector"]["harm"])

# Construct a clean, high-contrast matplotlib plot layout for forensic presentations
plt.figure(figsize=(10, 5))
plt.plot(steps, operational_radius, label="Operational Buffer (rm)", color="#3498db", marker="o", linewidth=2)
plt.plot(steps, harm_radius, label="Harm Reach (rh)", color="#e74c3c", marker="s", linewidth=2)

plt.title("ProvenanceShield Structural Trajectory Diagnostics", fontsize=14, fontweight="bold")
plt.xlabel("Forensic Timeline Steps", fontsize=12)
plt.ylabel("Manifold Metric Radii", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend(fontsize=11)
plt.show()
```
